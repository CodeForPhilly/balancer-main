# Assistant eval review

Load two result CSVs produced by `eval_assistant.py` (one per branch) and compare the
assistant's responses side by side.

Set `DEVELOP_CSV` and `FEATURE_CSV` below to the two files in `results/` you want to compare.

```python
import pandas as pd

DEVELOP_CSV = "results/develop-<timestamp>.csv"
FEATURE_CSV = "results/<branch>-<timestamp>.csv"

develop_df = pd.read_csv(DEVELOP_CSV)
feature_df = pd.read_csv(FEATURE_CSV)

# Outer join on question so questions missing from either run stay visible.
comparison = develop_df.merge(
    feature_df, on="question", how="outer", suffixes=("_develop", "_feature")
)
comparison[["question", "response_output_text_develop", "response_output_text_feature"]]
```
